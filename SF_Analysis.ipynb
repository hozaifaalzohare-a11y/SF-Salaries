Top 10 jops

```python
df['BasePay'] = pd.to_numeric(df['BasePay'], errors='coerce')

df['OvertimePay'] = pd.to_numeric(df['OvertimePay'], errors='coerce')

df['JobTitle'] = df['JobTitle'].str.strip().str.upper()

valid = df[(df['BasePay'] > 0) & (df['OvertimePay'].notna())].copy()

grp = valid.groupby('JobTitle').agg(
    Count=('Id', 'count'),
    Avg_BasePay=('BasePay', 'mean'),
    Avg_OvertimePay=('OvertimePay', 'mean'),
    Pct_With_OT=('OvertimePay', lambda x: (x > 0).mean() * 100)
).round(2)

grp['Avg_OT_Ratio_%'] = (grp['Avg_OvertimePay'] / grp['Avg_BasePay'] * 100).round(2)

grp_filtered = grp[grp['Count'] >= 30].sort_values('Avg_OT_Ratio_%', ascending=False)
grp_filtered.head(10)
```

```python
import matplotlib.pyplot as plt

top10 = grp_filtered.head(10).sort_values('Avg_OvertimePay')

plt.figure(figsize=(12, 6))

plt.barh(
    top10.index,
    top10['Avg_OvertimePay']
)

plt.xlabel('Average Overtime Pay')
plt.ylabel('Job Title')
plt.title('Top 10 Job Titles by Average Overtime Pay')

plt.tight_layout()
plt.show()
```

```python
top10_Overtime = df.groupby('JobTitle')['OvertimePay'].sum().sort_values(ascending=False).head(10)
top10_Overtime
```
